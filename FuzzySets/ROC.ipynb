# Fuzzy Inference

On this Jupyter page, lets look at the compositional rule of inference.

Consider Zadeh's compositional rule of inference 

Lets look at some code to calculate this (following the example from our fuzzy relation Jupyter page)

```python
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

# Reminder: These are the vector form of fuzzy set in there corresponding domain
# For example, A is SMALL. Domain is [1,2,3,4]
#              B is MEDIUM, domian is [a,b,c,d,e]
A = np.asarray( [1.0, 0.8, 0.0, 0.0] )
B = np.asarray( [0.0, 0.5, 1.0, 0.5, 0.0] )
```

```python
# (7.4) (7.5) (7.6) in textbook.
# reminder: The value we got from the implication is the element in the relation matrix.
# If Rz(2,d) = 0.9, that means "if A is 2, the B is mostly d".
def implication_zadeh(x,y):
    return min( 1.0 , 1.0 - x + y )

def implication_correlation_min(x,y):
    return min( x, y )

def implication_correlation_prod(x,y):
    return (x * y)
```

```python
# Initialize the Relation matrix. This matrix correspond to the set cross product between A and B. (all combinations)
R = np.zeros((len(A),len(B)))
for i in range(len(A)):
    for j in range(len(B)):
        R[i,j] = implication_zadeh(A[i],B[j])

print(R)
```

```python
newB = np.zeros(len(B))
for j in range(len(B)):
    for i in range(len(A)):
        temp = min(A[i],R[i,j])
        if(temp > newB[j]):
            newB[j] = temp
print(newB)
```

What if we try corr min?

```python
R = np.zeros((len(A),len(B)))
for i in range(len(A)):
    for j in range(len(B)):
        R[i,j] = implication_correlation_min(A[i],B[j])
        
newB = np.zeros(len(B))
for j in range(len(B)):
    for i in range(len(A)):
        temp = min(A[i],R[i,j])
        if(temp > newB[j]):
            newB[j] = temp
print(newB)
```
