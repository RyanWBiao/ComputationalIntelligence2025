# Backprop algorithm

Note, this is w.r.t. sum of squared error (SSE).

---

* Pick your MLP architecture
  
  * number of hidden layers, activation functions, etc.

* Training data
   
  * $T = \{ (\textbf{x}(k),\textbf{d}(k)) \}_{k=1}^{K}$
  
  * note: $\textbf{x}$ and $\textbf{d}$ are vectors and $(k)$ is one instance

* Init weights

  * e.g., random values

* Repeat
  
  * Do, for $k = 1,\ldots,K$ (samples in one epoch)
  
     * Forward computation (layer-by-layer)
     
       * Do, for $h = 1, 2, \ldots, $ to $H$
      
          * compute $v_j^h(k) = \left( \textbf{w}_j^h(k) \right)^t \textbf{y}^{h-1}(k)$
            
            * $j$th dot product at layer $h$
            
            * note: $\textbf{y}^0(k) = \textbf{x}(k)$
        
          * compute $\textbf{y}^h_j(k) = \varphi_j\left( v_j^h(k) \right)$
          
            * $j$th activation at layer $h$
            
       * End for (forward pass)
                
     * Error
       
       * compute $e_j^H(k) = d_j(k) - y_j^H(k)$

     * Backward       
     
       * Do, for $h = H, H-1, \ldots,$ to $1$
     
         * compute $\delta_j^h(k) = \left\{ \begin{array}{cc} 
                e^H_j(k) \varphi^{'}(v_j^H(k)) & \text{if at output layer } (h == H) \\
                \varphi^{'}(v_j^h(k)) \sum_{l=1}^{L}{ \delta_l^{h+1}(k)w_{lj}^{h+1}(k) } & \text{else} \\
                \end{array} \right.$
                
         * weight update 
     
            * $w_{ji}^h(k+1) = w_{ji}^h(k) + \alpha \Big[ \delta_j^h(k) y_i^{h-1}(k) \Big]$       
       
     * End for (backward pass) 
            
  * End for (epoch)
  
  * Shuffle the data
  
* Until DONE



---

# 📘 Backpropagation Algorithm — Human-Friendly Explanation

This section explains the standard **multilayer feedforward neural network backpropagation algorithm**.
For each training sample (k), we perform:

1. **Forward pass**
2. **Error computation**
3. **Backward pass (delta computation)**
4. **Weight update**

---

## 1️⃣ Loop Over Training Samples

For each epoch:

```
for k = 1 … K:
    forward pass
    compute error
    backward pass
    update weights
shuffle data
repeat until done
```

* (K): number of samples in the dataset
* (k): current sample index

---

## 2️⃣ Forward Computation (Layer by Layer)

For each layer (h = 1,2,\dots,H):

### 2.1 Linear Combination

[
v_j^h(k) = \left( \mathbf{w}_j^h(k) \right)^T \mathbf{y}^{,h-1}(k)
]

* (\mathbf{w}_j^h): weight vector of neuron (j) in layer (h)
* (\mathbf{y}^{h-1}(k)): outputs from previous layer
* For the first layer: (\mathbf{y}^0(k) = \mathbf{x}(k)) (input vector)

This computes the **net input** of neuron (j).

---

### 2.2 Activation

[
y_j^h(k) = \varphi\big(v_j^h(k)\big)
]

* (\varphi): activation function (Sigmoid, ReLU, etc.)

This produces the **output** of neuron (j).

After the forward pass we have:

* (v_j^h(k)): net inputs
* (y_j^h(k)): activations

These will be used in backprop.

---

## 3️⃣ Compute Output Error

At the output layer (H):

[
e_j^H(k) = d_j(k) - y_j^H(k)
]

* (d_j(k)): desired target
* (y_j^H(k)): network prediction

This error drives the backpropagation.

---

## 4️⃣ Backward Pass (Delta Computation)

Backprop runs from the output layer backward:

### 4.1 Delta at Output Layer

If (h = H):

[
\delta_j^H(k) = e_j^H(k),\varphi'!\left(v_j^H(k)\right)
]

* Output layer delta = error × derivative of activation

---

### 4.2 Delta at Hidden Layers

If (h < H):

[
\delta_j^h(k)
= \varphi'!\left(v_j^h(k)\right)
\sum_{l=1}^{L}
\delta_l^{h+1}(k), w_{lj}^{h+1}(k)
]

Meaning:

* Multiply derivative of activation at neuron (j)
* By the weighted sum of deltas from the next layer

This is the **chain rule** in action.

---

## 5️⃣ Weight Update Rule

For each weight connecting neuron (i) in layer (h-1)
to neuron (j) in layer (h):

[
w_{ji}^h(k+1)
= w_{ji}^h(k)

* \alpha,\delta_j^h(k), y_i^{h-1}(k)
  ]

- (\alpha): learning rate
- (\delta_j^h(k)): delta of current layer
- (y_i^{h-1}(k)): output of previous layer neuron (i)

This is gradient descent:

* Increase weight if the current input contributed positively
* Decrease otherwise

---

## 🧠 One-Sentence Summary

Backprop for each training sample:

1. **Forward:**
   [
   v = w^T y,\quad y = \varphi(v)
   ]

2. **Error:**
   [
   e = d - y
   ]

3. **Backward:**

   * Output layer: (\delta = e \cdot \varphi'(v))
   * Hidden layers: (\delta = \varphi'(v)\sum(\delta_{\text{next}} \cdot w))

4. **Update:**
   [
   w \leftarrow w + \alpha ,\delta,y_{\text{prev}}
   ]
